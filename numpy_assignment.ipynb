```python
#10. Calculate mean of each feature using axis=0


import numpy as np

X = np.array([[10, 20, 30],
              [40, 50, 60],
              [70, 80, 90]])

print(np.mean(X, axis=0))
```

```python
#11. Filter rows where first feature is greater than 50

import numpy as np

X = np.array([[20, 30, 40],
              [60, 70, 80],
              [90, 50, 30],
              [40, 80, 20]])

result = X[X[:, 0] > 50]

print(result)
```

```python
#12. Standardize every feature column

import numpy as np

X = np.array([[10, 20],
              [20, 40],
              [30, 60]], dtype=float)

standardized = (X - np.mean(X, axis=0)) / np.std(X, axis=0)

print(standardized)
```

```python
#20. Difference between A * B and A @ B

import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

print("A * B:")
print(A * B)

print("A @ B:")
print(A @ B)
```

```python
#Normalization and Standardization


import numpy as np

data = np.array([10, 20, 30, 40, 50], dtype=float)

normalized = (data - np.min(data)) / (np.max(data) - np.min(data))
standardized = (data - np.mean(data)) / np.std(data)

print("Original:", data)
print("Normalized:", normalized)
print("Standardized:", standardized)
print("Standardized mean:", np.mean(standardized))
print("Standardized std:", np.std(standardized))
```
